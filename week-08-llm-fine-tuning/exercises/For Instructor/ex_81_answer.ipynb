## **2. multilingual translation**

Instructions:
* Load a multilingual translation model with the pipeline() API.
* Choose one sentence in any language (e.g., English or Korean).
* Select 3–4 target languages (e.g., French, Japanese, Spanish, German).
* Translate the same sentence into each selected language.
* Compare the translations and observe differences in style or wording.

```
# Load a multilingual translation model with the pipeline() API.
translator = pipeline("translation", model="facebook/mbart-large-50-many-to-many-mmt")
```

```
# Choose one sentence in any language.
text = "Artificial intelligence is changing the world."
```

```
# Select 3–4 target languages (e.g., French, Japanese, Spanish, German).
targets = ["fr_XX", "es_XX", "de_XX", "ja_XX"]

# Translate the same sentence into each selected language.
for tgt in targets:
    result = translator(text, src_lang="en_XX", tgt_lang=tgt)
    print(tgt, ":", result[0]["translation_text"])
```
