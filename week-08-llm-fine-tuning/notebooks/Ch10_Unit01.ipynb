# **pipeline API**

## **1. import library**

```
from transformers import pipeline
```

## **2. using pipeline**

### (1) basic use pattern

```
pipe = pipeline("<task_name>", model="<model_name_optional>")
result = pipe("<your input text>")
print(result)
```

### (2) Sentiment Analysis

```
# Create a sentiment-analysis pipeline
sentiment_pipe = pipeline("sentiment-analysis")

# Run on a single sentence
result = sentiment_pipe("I really enjoy learning about deep learning!")
print(result)
```

* Method 2

```
sentiment_pipe = pipeline("sentiment-analysis", model="bert-base-multilingual-cased")

text = "The user interface is beautiful, but the app is very slow."
result = sentiment_pipe(text)
print(result)
```

### (3) Zero-shot classification

```
# Create a zero-shot classification pipeline
zero_shot_pipe = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

text = "The user interface is beautiful, but the app is very slow."
candidate_labels = ["positive", "negative", "performance issue", "design"]

result = zero_shot_pipe(text, candidate_labels=candidate_labels)
print(result)
```

### (4) Translation

```
# Create a translation pipeline (English → French)
translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")

# Prepare an English sentence
text = "Artificial intelligence is transforming the world."

# Run the pipeline to translate into French
result = translator(text)

# Print the translated output
print(result)
```

```
translator = pipeline("translation", model="facebook/mbart-large-50-many-to-many-mmt")

text = "Artificial intelligence is changing the world."
result = translator(text, src_lang="en_XX", tgt_lang="fr_XX") # English → French
print(result)
```
