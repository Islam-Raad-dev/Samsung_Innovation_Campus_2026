### 1. Fetch live news:

```python
import requests
import re
```

```python
# Fetch live data (Space/Tech news)
url = "https://api.spaceflightnewsapi.net/v4/articles/?limit=50"
data = requests.get(url).json()
my_news = [article['title'] for article in data['results']]
```

### 2. Pre-processing news titles:

```python
# You can add more common words here.
STOPWORDS = {"a", "an", "the", "and", "or", "in", "on", "at", "to", "for", "with", "is", "of", "it"}

my_news_new = []
for text in my_news:
    # Basic Cleaning
    text = text.lower()
    text = re.sub(r"https?://\S+", "", text)
    text = re.sub(r"it's", "it is", text)
    text = re.sub(r"i'm", "i am", text)
    text = re.sub(r"\W", " ", text) 
    text = re.sub(r"\d", " ", text)
    
    # Tokenize by splitting on whitespace.
    words = text.split() 
    
    # Filter out stopwords and single letters.
    filtered_words = [w for w in words if w not in STOPWORDS and len(w) > 1]
    
    # Join back into a string.
    clean_text = ' '.join(filtered_words)
    my_news_new.append(clean_text)
```
