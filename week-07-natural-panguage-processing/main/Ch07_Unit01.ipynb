```python
import requests as rq
url = "https://en.wikipedia.org/wiki/Machine_learning"
header = {"User-Agent": (
          "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
          "AppleWebKit/537.36 (KHTML, like Gecko) "
          "Chrome/129.0 Safari/537.36")
}
res = rq.get(url, headers=header)
print(res.status_code)
print(res.text[:1000])
```

```python
import requests, bs4
url = "https://en.wikipedia.org/wiki/Machine_learning"
header = {"User-Agent": (
          "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
          "AppleWebKit/537.36 (KHTML, like Gecko) "
          "Chrome/129.0 Safari/537.36")
}
res = requests.get(url, headers=header)
soup = bs4.BeautifulSoup(res.text, 'html.parser')
x = soup.find_all('p')
text = ''
for i in range(len(x)):
    text += x[i].text.strip() + '\n'
print(text[:1200])
```

## 1.3. String Manipulation

```python
import re
my_regex = re.compile("([0-9]+)[^0-9]+([0-9]+)")
m = my_regex.search("Anna is 15 years old and John is 12 years old.")
print(m.group(0))
print(m.group(1))
print(m.group(2))
```

```python
my_regex = re.compile(r"(\D+)(\d{3})-(\d{4})-(\d{4})")
masked = my_regex.sub(r"\1\2-****-\4", "John 010-1234-5678")
print(masked)
```

```python
my_regex = re.compile("(\D+)((\d+)\D+(\d+)\D+(\d+))")
m = my_regex.search("John 010-1234-5678")
print("Phone number: " + m.group(2))
```

## 1.4. User Defined Function

```python
def grep(reg_exp, my_text):
    result = []
    reg_exp_compiled = re.compile(reg_exp)
    for i in range(len(my_text)):
        if reg_exp_compiled.search(my_text[i]):
            result.append(my_text[i])
    return result
```

```python
# Find strings starting with numbers
grep("^[0-9]", ["100 is my score.", "I am 21 years old."])
```
