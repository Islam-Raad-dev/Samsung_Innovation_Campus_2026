```python
import tensorflow_datasets as tfds
import numpy as np
```

```python
# Hyperparameter definitions.
BUFFER_SIZE = 10000 
BATCH_SIZE = 64
NUM_EPOCHS = 20
```

```python
# Calculate the number of steps per epoch (60,000 images total).
steps_per_epoch = int(np.ceil(60000 / BATCH_SIZE))
```

```python
# Load MNIST dataset.
(ds_train, ds_test), ds_info = tfds.load(
    'mnist',
    split=['train', 'test'],
    as_supervised=True, 
    with_info=True
)
```

```python
# Preprocessing function.
def preprocess(image, label):
    image = tf.image.convert_image_dtype(image, tf.float32)
    return image, label
```

```python
# Build input pipeline.
# Training dataset
train_ds = (
    ds_train
    .map(preprocess)
    .shuffle(BUFFER_SIZE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# Test dataset.
test_ds = (
    ds_test
    .map(preprocess) 
    .batch(BATCH_SIZE) 
    .prefetch(tf.data.AUTOTUNE)
)
```

```python
# Define the DNN classifier model.
dnn_classifier = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28, 1)), 

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(16, activation='relu'),

    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile the model.
dnn_classifier.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy', 
    metrics=['accuracy']
)
```

```python
# Train the model.
dnn_classifier.fit(
    train_ds,
    epochs=NUM_EPOCHS,
    steps_per_epoch=steps_per_epoch,
    validation_data=test_ds 
)
```

```python
# Evaluate the model.
eval_loss, eval_acc = dnn_classifier.evaluate(test_ds)
print("Evaluation loss:", eval_loss)
print("Evaluation accuracy:", eval_acc)
```
