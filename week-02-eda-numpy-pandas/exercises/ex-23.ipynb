### 1. Data Feature Engineering:

#### 1.1. Add Columns:

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('TV_Sales.csv')
df.head()
```

```python
df['price_diff'] = df['price'] - df['comp_price']
df.head()
```

```python
df['price_tot'] = df['price'] * df['sales']
df.head()
```

#### 1.2. Data Type Conversion:

```python
df.dtypes
```

```python
df['price_tot'] = df['price_tot'].astype(int)
```

```python
df.dtypes
```

#### 1.3. Categorical Data Conversion:

```python
count, bin_dividers = np.histogram(df['sales'], bins=3)
print(bin_dividers)
```

```python
bin_names = ['Low Sales', 'Normal Sales', 'High Sales']

df['sales_bin'] = pd.cut(x=df['sales'],
                     bins=bin_dividers,
                     labels=bin_names,
                     include_lowest=True)

df.head()
```

#### 1.4. One-Hot Vector:

```python
sales_dummies = pd.get_dummies(df['sales_bin']).astype('int')
sales_dummies.head()
```

#### 1.5. Normalization:

```python
df['sales'].describe()
```

```python
df['sales'] = (df['sales'] - df['sales'].min()) / (df['sales'].max() - df['sales'].min())
df['sales'].describe()
```

```python
# Process multiple columns at once:
cols = ['comp_price', 'income', 'price', 'price_diff', 'price_tot']
df[cols].describe()
```

```python
df[cols] = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())
df[cols].describe()
```
