### 1. Classification with Tree:

```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import warnings
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn import metrics, preprocessing
warnings.filterwarnings(action='ignore')                  # Turn off the warnings.
%matplotlib inline
```

#### 1.1. Read in data:

```python
# Go to the directory where the data file is located. 
# os.chdir(r'~~')                # Please, replace the path with your own.   
```

```python
# Read the already pre-processed data.
df = pd.read_csv('data_titanic_2.csv', header='infer')
```

```python
df.shape
```

```python
df.head(3)
```

```python
X = df.drop(columns=['Survived'])
Y = df['Survived']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1234)
```

#### 1.2. Tree hyperparameter optimization:

```python
depth_grid = range(1, 21)
min_samples_leaf_grid = range(10, 31)
max_leaf_nodes_grid = range(2, 21)
parameters = {'max_depth': depth_grid, 
              'min_samples_leaf': min_samples_leaf_grid, 
              'max_leaf_nodes': max_leaf_nodes_grid}
```

```python
gridCV = GridSearchCV(DecisionTreeClassifier(), 
                      parameters, 
                      cv=10, 
                      n_jobs=-1)
gridCV.fit(X_train, Y_train);
best_depth = gridCV.best_params_['max_depth']
best_min_samples_leaf = gridCV.best_params_['min_samples_leaf']
best_max_leaf_nodes = gridCV.best_params_['max_leaf_nodes']
```

```python
print("Tree best depth:", best_depth)
print("Tree best min_samples_leaf:", best_min_samples_leaf)
print("Tree best max_leaf_nodes:", best_max_leaf_nodes)
```

```python
DTC_best = DecisionTreeClassifier(max_depth=best_depth,
                                  min_samples_leaf=best_min_samples_leaf,
                                  max_leaf_nodes=best_max_leaf_nodes)
DTC_best.fit(X_train, Y_train);
Y_pred = DTC_best.predict(X_test)
print( "Tree best accuracy: ", round(metrics.accuracy_score(Y_test, Y_pred), 3))
```

### 2. Regression with Tree:

#### 2.1. Read in data: 

```python
df = pd.read_csv('TV_Sales.csv')
```

```python
df.head(5)
```

```python
# Pair-wise correlation matrix.
round(df.corr(numeric_only=True), 2)
```

```python
X = df.drop(columns=['sales'])
Y = df['sales']
```

```python
# Convert into dummy variables.
X = pd.get_dummies(X, columns=['shelve_loc', 'city', 'local'], drop_first=True, dtype=int)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1234)
```

#### 2.2. Tree hyperparameter optimization:

```python
depth_grid = range(1, 21)
min_samples_leaf_grid = range(10, 31)
max_leaf_nodes_grid = range(2, 21)
parameters = {'max_depth': depth_grid, 
              'min_samples_leaf': min_samples_leaf_grid, 
              'max_leaf_nodes': max_leaf_nodes_grid}
```

```python
gridCV = GridSearchCV(DecisionTreeRegressor(), 
                      parameters, 
                      cv=10, 
                      n_jobs = -1)  
gridCV.fit(X_train, Y_train)
best_depth = gridCV.best_params_['max_depth']
best_min_samples_leaf = gridCV.best_params_['min_samples_leaf']
best_max_leaf_nodes = gridCV.best_params_['max_leaf_nodes']
```

```python
print("Tree best depth:", best_depth)
print("Tree best min_samples_leaf:", best_min_samples_leaf)
print("Tree best max_leaf_nodes:", best_max_leaf_nodes)
```

```python
DTR_best = DecisionTreeRegressor(max_depth=best_depth,
                                 min_samples_leaf=best_min_samples_leaf,
                                 max_leaf_nodes=best_max_leaf_nodes)
DTR_best.fit(X_train, Y_train)
Y_pred = DTR_best.predict(X_test)
print("Tree best RMSE:", round(metrics.root_mean_squared_error(Y_test, Y_pred), 3))
```
