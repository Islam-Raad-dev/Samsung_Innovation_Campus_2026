### 1. Linear regression diagnostics and modeling using StatsModels library:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn import metrics
%matplotlib inline
```

#### 1.1. Read in the data:

```python
df = pd.read_csv('TV_Sales.csv')
```

```python
# Convert into dummy variables.
df = pd.get_dummies(df, columns=['shelve_loc', 'city', 'local'], drop_first=True, dtype=int)
```

```python
# Separate into X and Y.
X = df.drop(columns=['sales'])
Y = df['sales']
```

```python
df.head()
```

#### 1.2. Linear modeling using R-styled formula:

```python
# Using all the explanatory variables.
myModel1 = smf.ols(formula="sales ~ comp_price + income + price + shelve_loc_Good + shelve_loc_Medium + city_Yes + local_Yes ", data=df)
result1 = myModel1.fit()
result1.summary()
```

**NOTE: city_Yes does not seem to be significant. Adjusted $R^2 = 0.752$ and $AIC = 1416$.** 

```python
# Dropping the explanatory variables 'city_YES'.
myModel2 = smf.ols(formula="sales ~ comp_price + income + price + shelve_loc_Good + shelve_loc_Medium + local_Yes ", data=df)
result2 = myModel2.fit()
result2.summary()
```

**NOTE: All the explanatory variables are significant. Adjusted $R^2 = 0.752$ and $AIC = 1414$.** 

**CONCLUSION: The second model is slightly optimized compared to the first one.**
