### 1. Linear regression and diagnostics:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn import metrics
%matplotlib inline
```

#### 1.1. Read in the data:

```python
df = pd.read_csv('TV_Sales.csv')
```

```python
# Check the top data.
df.head()
```

```python
# Separate into X and Y.
X = df.drop(columns=['sales'])
Y = df['sales']
```

#### 1.2. Data exploration:

```python
df.tail(5)
```

```python
# Descriptive statistics of the variables (columns).
df.describe()
```

```python
# Pair-wise correlation matrix.
np.round(df.corr(numeric_only=True),2)
```

```python
# Visualize the correlation matrix.
sns.heatmap(df.corr(numeric_only=True),cmap='coolwarm')
plt.show()
```

```python
# Visualize price vs sales.
plt.scatter(X['price'], Y, c='g', s=15, alpha=0.5)
plt.xlabel('price')
plt.ylabel('sales')
plt.show()
```

#### 1.3. Feature engineering:

```python
# Convert into dummy variables.
X = pd.get_dummies(X, columns=['shelve_loc', 'city', 'local'], drop_first=True, dtype=int)
```

#### 1.4. Train by linear regression:

```python
# Train.
lm = LinearRegression(fit_intercept=True)
lm.fit(X, Y)
```

```python
# The intercept.
lm.intercept_
```

```python
# The rest of coefficients (parameters).
lm.coef_
```

```python
# Display the parameters as a DataFrame.
parametersDF = pd.DataFrame(lm.coef_.reshape(1, 7), index=['Parameter Value'], columns=list(X))
parametersDF['Intercept'] = lm.intercept_
parametersDF
```

#### 1.5. Diagnostics:

```python
# In-sample prediction.
predY = lm.predict(X)
```

```python
# Display real Y vs predicted Y.
plt.scatter(Y,predY,c = 'blue', s=15, alpha=0.5)
plt.xlabel('REAL SALES')
plt.ylabel('PREDICTED SALES')
plt.show()
```

```python
Y
```

```python
# Calculate the correlation between the real Y and predicted Y.
pd.Series(Y.corr(pd.Series(predY)))
```

```python
# Coefficient of determination (R^2):
lm.score(X, Y)
```

#### 1.6. In-sample and out-of-sample testing:

```python
# Split the dataset.
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=123)
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
# predY_in = in-sample prediction of Y.
# predY_out = out-of-sample prediction of Y.
lm = LinearRegression()
lm.fit(X_train,Y_train)
Y_pred_train = lm.predict(X_train)
Y_pred_test = lm.predict(X_test)
```

```python
print('In-sample MSE is      :', metrics.mean_squared_error(Y_train, Y_pred_train))
print('Out-of-sample MSE is  :', metrics.mean_squared_error(Y_test, Y_pred_test))
print('-' * 43)
print('In-sample RMSE is     :', metrics.root_mean_squared_error(Y_train, Y_pred_train))
print('Out-of-sample RMSE is :', metrics.root_mean_squared_error(Y_test, Y_pred_test))
```

NOTE: In-sample error are a bit smaller but not by much.

#### 1.7. Residual analysis:

```python
# Calculate residual.
residual = Y_train - Y_pred_train
```

```python
# Q: Can you check "visually" that the mean = 0 and variance = constant?
plt.scatter(Y_train, residual, c='red', s=15, alpha=0.5)
plt.xlabel('Y')
plt.ylabel('Residual')
plt.title('Residual')
plt.show()
```

```python
# Q: Are the residuals normally distributed centered around 0?
sns.histplot(residual, bins=50, color='green').set_title("Residual Histogram")
plt.show()
```
